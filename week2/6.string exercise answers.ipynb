### Exercise: Simple Personalized Message and Age Calculation

#### Objective:
Create a Python program that takes user input for their name and birth year. Then, calculate their age and generate a personalized message displaying their name and age.

#### Steps:
1. Ask the user to input their name.
2. Ask the user to input their birth year.
3. Calculate the user's age based on the current year (you may use the `input()` function to ask the user for the current year).
4. Display a personalized message containing the user's name and calculated age.

#### Example Output:
```
Please enter your name: Bob
Please enter your birth year: 1995
Please enter the current year: 2023
Hello, Bob! You are 28 years old.
```

#### Exercise Code Template:

```python
# Step 1: Ask for user input - name
name = input("Please enter your name: ")

# Step 2: Ask for user input - birth year
birth_year = int(input("Please enter your birth year: "))

# Step 3: Ask for user input - current year
current_year = int(input("Please enter the current year: "))

# Step 4: Calculate age
age = current_year - birth_year

# Step 5: Display personalized message
print(f"Hello, {name}! You are {age} years old.")
```

#### Exercise: String Manipulation and Analysis

#### Objective:
Create a Python program that accepts a sentence from the user, performs various string manipulations and analysis, and displays the results.

#### Steps:
1. Ask the user to input a sentence.
2. Strip the leading and trailing whitespace from the sentence.
3. Display the length of the cleaned sentence (number of characters).
4. Convert the cleaned sentence to uppercase and display it.
5. Split the cleaned sentence into words and display them on separate lines.
6. Count the occurrences of a specific letter (e.g., 's') in the cleaned sentence and display the count.

#### Example Output:
```
Please enter a sentence:   This is a sample sentence.   
Length of the sentence: 25 characters
Uppercase Sentence: THIS IS A SAMPLE SENTENCE.
Words in the Sentence:
[This,is,a,sample,sentence.]
Occurrences of 's': 3
```

####  Exercise Code Template:

```python
# Step 1: Ask for user input - sentence and strip leading/trailing whitespace
sentence = input("Please enter a sentence: ").strip()

# Step 2: Display the length of the cleaned sentence
length = len(sentence)
print(f"Length of the sentence: {length} characters")

# Step 3: Convert the cleaned sentence to uppercase and display
uppercase_sentence = sentence.upper()
print(f"Uppercase Sentence: {uppercase_sentence}")

# Step 4: Split the cleaned sentence into words and display each word
words = sentence.split()
print("Words in the Sentence:")
print(words)  # Use join() to display words on separate lines

# Step 5: Count the occurrences of a specific letter ('s') and display count
letter = 's'  # Change this letter to count occurrences of a different letter
letter_count = sentence.count(letter)
print(f"Occurrences of '{letter}': {letter_count}")
```
