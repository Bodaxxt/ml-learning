# Use Case Parsing and Storing Multi-Line String Data  
**Objective:** Illustrate parsing a multi-line string containing employee information and storing it in separate lists for names, salaries, and teams.

```python
# Multi-line string data example
data = """Alice, 50000, Sales
Bob, 60000, Marketing
Charlie, 55000, Product
David, 52000, Sales
Eva, 58000, Marketing"""

# Initialize empty lists
names = []
salaries = []
teams = []

# Split the data into lines
lines = data.split("\n")

# Process each line
for line in lines:
    # Split each line into individual data points
    individual_data = line.split(", ")
    
    # Assign values to respective lists
    names.append(individual_data[0])
    salaries.append(int(individual_data[1]))  # Convert salary to integer
    teams.append(individual_data[2])

# Display the extracted data
print("Names:", names)
print("Salaries:", salaries)
print("Teams:", teams)
```

This Python code parses a multi-line string `data` containing structured employee information. It separates this data into lines, then further breaks down each line into individual data points using `split()` based on the comma separator. The code then assigns these values to respective lists (`names`, `salaries`, `teams`). Finally, it displays the extracted data from these lists. This example demonstrates the step-by-step process of parsing structured data from a string and storing it into separate lists for efficient management and manipulation.

# Use Case 2 Counting Products and Extracting Unique Items  
**Objective:** Illustrate creating a set of unique values and a dictionary to count occurrences from a list of 20 products.

```python
# List of 20 products
products = [
    "T-shirt", "Shorts", "Socks", "T-shirt", "Shoes", "Shorts", "Socks", "Cap",
    "Trousers", "T-shirt", "Shoes", "Shorts", "Socks", "T-shirt", "Shoes", "Socks",
    "Cap", "Trousers", "T-shirt", "Shoes"
]

# Creating a set to extract unique items
unique_items = set(products)

# Creating a dictionary to count occurrences of each item
product_count = {}
for product in products:
    if product in product_count:
        product_count[product] += 1
    else:
        product_count[product] = 1

# Displaying unique items and their count
print("Unique Items:", unique_items)
print("Product Count:", product_count)
```

This Python code first initializes a list `products` containing 20 different items. It then creates a `set` called `unique_items` to extract unique values from the list, removing duplicates. Following that, it creates a `dictionary` named `product_count` to count the occurrences of each item within the list using a `for` loop and conditional statements. Finally, it displays the unique items and their respective count using `print()` statements. This example demonstrates the extraction of unique values and the counting of occurrences from a list using sets and dictionaries, respectively.
