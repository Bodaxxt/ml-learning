## Example 4
implement a program with the following aspects

1. **Receive Input:**
   - Use `input()` to receive a string input from the user.

2. **Strip Whitespace:**
   - Apply the `strip()` method to remove leading and trailing whitespace from the entered string.

3. **Check for Space:**
   - Utilize the `find()` method to search for a space character within the stripped string.
   - If `find()` returns `-1`, indicate that there's only one word in the string.
   - If `find()` returns a non-negative value, the string likely contains more than one word.

4. **Count Words (if multiple words detected):**
   - Use the `split()` method to split the string into words.
   - Determine the number of words using `len()` on the resulting list.

5. **Print Results:**
   - Display "One word only" if `find()` returns `-1`.
   - If `find()` returns a non-negative value, print the count of words detected.

By following these steps, users can determine if a string contains more than one word by inspecting the presence of spaces and subsequently counting the words if multiple words are present.

````python
String = input("Please, Enter the message: ").strip()
Index = String.find(" ")
if Index == -1 :
    print("only one word in the string")
else:
    Num_of_Words = len(String.split(" "))
    print(f"String contains more than one word and the number of words are {Num_of_Words}")
````

## example 5
here are the steps for creating a program that performs various operations on a predefined list based on user input:

1. **Define the List:**
   - Assign the list `[1, 5, 9, 20, 8, 3]` to a variable.

2. **Receive User Input:**
   - Use `input()` to prompt the user to enter an operation (e.g., sum, number of items, max, min, sort, average).

3. **Perform Operations:**
   - Use conditional statements (`if`, `elif`) to check the user input.
   - For each input, execute the corresponding operation:
     - If the input is 'sum', calculate the sum of the list.
     - If the input is 'number of items', use the `len()` function to return the number of items.
     - If the input is 'max', find and display the maximum value in the list.
     - If the input is 'min', find and display the minimum value in the list.
     - If the input is 'sort', use the `sorted()` method to sort the list.
     - If the input is 'average', calculate and print the average of the list.

4. **Handle Invalid Inputs:**
   - Account for invalid inputs or unexpected user entries.

````python
li = [1, 5, 9, 20, 8, 3]
print(' 1 : "Sum",\n 2 : "Number of items",\n 3 : "Max",\n 4 : "Min",\n 5 : "Sort",\n 6 : "Average"')
opertion = int(input("Please, Enter the number of opertion: "))
Methods = {
    1 : "Sum",
    2 : "Number of items",
    3 : "Max",
    4 : "Min", 
    5 : "Sort",
    6 : "Average",
    opertion : "In Valid Option"
}

if Methods[opertion] == "Sum":
    print(sum(li))
elif Methods[opertion] == "Number of items":
    print(len(li))
elif Methods[opertion] == "Max":
    print(max(li))
elif Methods[opertion] == "Min":
    print(min(li))
elif Methods[opertion] == "Sort":
    print(sorted(li))
elif Methods[opertion] == "Average":
    print(sum(li) / len(li))
else:
    print(Methods[opertion])
````

# Store comparsions Solution

````python
# Revenue lists for Store A and Store B for a week (Friday to Thursday)
revenue_store_A = [1200, 1300, 1100, 1450, 900, 1600, 1750]  # Revenue for Store A
revenue_store_B = [1100, 1350, 1000, 1400, 950, 1650, 1800]  # Revenue
#Day_of_week = ["Fri", "Sat", "Sun", "Mon", "Tues","Wed", "Thru"]
Day_of_week = {
    0 : "Fri",
    1 : "Sat",
    2 : "Sun",
    3 : "Mon",
    4 : "Tues",
    5 : "Wed",
    6 : "Thru"
}
````

````python
Sum_A = sum(revenue_store_A) 
Sum_B = sum(revenue_store_B)
````

````python
if Sum_A > Sum_B:
    print("Total Revenue for Store A is bigger then Store B")
    Index = revenue_store_A.index(max(revenue_store_A))
    Best_Sales_Day = Day_of_week[Index]
    print(f"The Best Day of Sales is {Best_Sales_Day}")
elif Sum_A < Sum_B:
    print("Total Revenue for Store A is less then Store B")
    Index = revenue_store_B.index(max(revenue_store_B))
    Best_Sales_Day = Day_of_week[Index]
    print(f"The Best Day of Sales is {Best_Sales_Day}")
else:
    print("Total Revenue for Store A is equal Store B")
    
````
