```python
print("Simple Calculator")
while True:

    print("1. Add")
    print("2. Subtract")
    print("3. Multiply")
    print("4. Divide")
    print('6. cummilative')
    print("5. Exit")

    choice = input("Enter your choice (1-5): ")

    if choice=='5':
        break

    num1 = float(input("Enter the first number: "))
    num2 = float(input("Enter the second number: "))

    if choice == '1':
        result = num1 + num2
        print("Result: ", result)
    elif choice == '2':
        result = num1 - num2
        print("Result: ", result)
    elif choice == '3':
        result = num1 * num2
        print("Result: ", result)
    elif choice == '4':
        if num2 != 0:
            result = num1 / num2
            print("Result: ", result)
        else:
            print("Error! Division by zero is not allowed.")
    else:
        print("Invalid choice! Please enter a valid option.")
    
    

    with open('result.txt','a') as file:
        operations=['+','-','x','/']
        oper=operations[int(choice) -1]
        file.write(f'{num1} {oper} {num2} = {result}\n')
```

```python
import random
computer_score=0
user_score=0
while  computer_score ==3 or user_score==3:
    print("Rock-Paper-Scissors")
    print("1. Rock")
    print("2. Paper")
    print("3. Scissors")
    print("4. Exit")

    choice = input("Enter your choice (1-4): ")

    player_choice = int(choice)

    if player_choice < 1 or player_choice > 3:
        print("Invalid choice! Please enter a valid option.")
        continue



    computer_choice = random.randint(1, 3)

    print("Player chooses:", player_choice)
    print("Computer chooses:", computer_choice)

    if player_choice == computer_choice:
        print("It's a tie!")
    elif (
        (player_choice == 1 and computer_choice == 3)
        or (player_choice == 2 and computer_choice == 1)
        or (player_choice == 3 and computer_choice == 2)
    ):
        print("Player wins!")
        user_score+=1
    else:
        print("Computer wins!")
        computer_score+=1
    
    print(f'the user score = {user_score} and computer_score ={computer_score}')
    with open ('rps.txt','a') as file:
        file.write(f'the user score = {user_score} and computer_score ={computer_score}')


```
