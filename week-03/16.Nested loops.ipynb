# Exploring Nested Loops

Welcome to the world of nested loops! In programming, loops play a vital role in repeating tasks efficiently. Nested loops take this concept further by placing one loop inside another. They are like Russian nesting dolls, where one doll holds another within it. Similarly, nested loops allow us to perform repetitive actions within other repetitive actions. This notebook will guide you through the fascinating world of nested loops, demonstrating how they can be used to solve complex problems by iterating over data structures in multiple layers.



### Nested For Loops

In this section, we'll explore the concept of nested loops by using nested `for` loops. Nested loops allow us to perform repetitive tasks within repetitive tasks, executing code for each combination of elements within a structure. Here's an example using nested `for` loops:

```python
for i in range(3):
    for j in range(2):
        print(f"({i}, {j})")
```

This code snippet demonstrates a nested `for` loop structure. The outer loop (`for i in range(3)`) runs three times, and for each iteration of the outer loop, the inner loop (`for j in range(2)`) runs twice. The output will display pairs of numbers, representing combinations of `i` and `j`.

### Analyzing Sales Data

Let's simulate a scenario where we have a list of sales records, each containing the sales figures for various products over a week. We'll use nested loops to calculate the total sales for each day and overall sales for the week.

```python
# Sales data for the week
sales_data = [
    [120, 150, 180],  # Day 1: Sales for Product A, B, C
    [100, 130, 140],  # Day 2
    [80, 120, 160],   # Day 3
    [200, 180, 150],  # Day 4
    [150, 100, 90]    # Day 5
]

# Calculate daily and weekly sales
total_weekly_sales = 0

for day, sales in enumerate(sales_data, start=1):
    total_daily_sales = sum(sales)
    total_weekly_sales += total_daily_sales

    print(f"Day {day}: Total Sales - {total_daily_sales}")

print(f"\nTotal Sales for the Week: {total_weekly_sales}")
```

**Explanation:**

- The `sales_data` list contains sales figures for different products over five days (nested lists represent each day's sales).
- We use a nested `for` loop to iterate through each day's sales figures.
- The `enumerate()` function helps us track the day number.
- For each day, the inner loop calculates the total sales by summing up the sales figures for each product.
- The total daily sales are printed for each day, and at the end, the total sales for the entire week are displayed.

````python
## try it by yourself...

sales_data = [
    [120, 150, 180],  # Day 1: Sales for Product A, B, C
    [100, 130, 140],  # Day 2
    [80, 120, 160],   # Day 3
    [200, 180, 150],  # Day 4
    [150, 100, 90]    # Day 5
]

total_weekly_sales=0

for day,value in enumerate(sales_data,start=1):
    total_sales_per_day=0
    for product in value:
        total_sales_per_day+=product
        
    total_weekly_sales=total_weekly_sales+total_sales_per_day
    print(f'the total sales for day {day} equal {total_sales_per_day}')
print(f'the total weekly sales {total_weekly_sales}')
````

### Inventory Analysis

Suppose we have a list of products and their quantities in stock. We'll use nested loops along with `if` conditions to identify products that are running low in stock.

]
